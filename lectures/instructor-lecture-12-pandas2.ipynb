## Group by

Group records that have the same value for a column

```python
data.groupby('Facility ID')
```

We can then apply an aggregating function to these groups.

```python
# note: make sure mean() is only applied to columns with numerical values, or you will get an error
data.groupby('Facility ID').mean(numeric_only=True).sort_values(by='Total Charges')
```

## Combining DataFrames

`pd.concat` concatenates rows (i.e., default axis=0) while merging columns with the same name

```python
df1 = pd.DataFrame(
    {
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
        "C": ["C0", "C1", "C2", "C3"],
        "D": ["D0", "D1", "D2", "D3"],
    },
    index=[0, 1, 2, 3],
)

df2 = pd.DataFrame(
    {
        "B": ["B2", "B3", "B6", "B7"],
        "D": ["D2", "D3", "D6", "D7"],
        "F": ["F2", "F3", "F6", "F7"],
    },
    index=[2, 3, 6, 7],
)
```

## Join

You can **join** two tables **on** a specific column (or columns)  
Rows that has the same value (or _key_) in that column will be combined  

```python
left = pd.DataFrame({"key": ["K0", "K1", "K2", "K3"],
                     "A": ["A0", "A1", "A2", "A3"],
                     "B": ["B0", "B1", "B2", "B3"],})


right = pd.DataFrame({"key": [ "K1", "K2", "K4"],
                      "C": ["C1", "C2", "C4"],
                      "D": ["D1", "D2", "D4"],})
```

```python
left
```

```python
right
```

## Inner join

Key must exist in both tables

```python
pd.merge(left,right,on='key',how='inner')
```

## Outer join

Key can exist in either table

```python
pd.merge(left,right,on='key',how='outer')
```

# Left (right) join

Key must exist in left (right) table

```python
pd.merge(left, right, on='key', how='left')
```

```python
pd.merge(left, right, on='key', how='right')
```

Train a predictive model for cost.

```python
X = procrows.drop(['Total Charges'],axis=1)
Y = procrows['Total Charges']
```

```python
import sklearn
from sklearn import tree, ensemble,preprocessing, feature_extraction
```

```python
X_dict = list(X.astype(np.str).T.to_dict().values())
```

```python
X_dict
```

```python
dvec = feature_extraction.DictVectorizer()
vect = dvec.fit_transform(X_dict)
```

```python
print(vect[0])
```

```python
dvec.get_feature_names()[793]
```

```python
from sklearn import metrics
```

```python
m = tree.DecisionTreeRegressor()
dtree = m.fit(vect,Y)
metrics.r2_score(dtree.predict(vect),Y)
```

```python
plt.plot(dtree.predict(vect),Y,'o',alpha=.1)
plt.xlabel("Prediction")
plt.ylabel("Actual")
```

```python
from sklearn import model_selection

gm = model_selection.GridSearchCV(m, {'max_depth':[None,2,3,4,6,8,10,15,20]},scoring='r2')
gres = gm.fit(vect,Y)
```

```python
gres.best_score_
```

```python
gres.best_params_
```
