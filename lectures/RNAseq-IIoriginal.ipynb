```python
# Jupyter Notebook Cell 1: Introduction and Setup

"""
Tutorial: Systematic Gene Expression Analysis for Healthy vs Apoptotic Cells in Python

This notebook demonstrates a full workflow:
- Simulating a gene expression count matrix for healthy vs apoptotic cells
- Quality control
- Variance stabilization and normalization (log, quantile, TMM, etc.)
- Differential expression analysis (using e.g., statsmodels, DESeq2-like workflow)
- Visualization: Volcano plot and heatmap

References:
- Cell death gene sets: Nature Communications s41419-024-06632-7, May 2024 [web:1]
- Analysis best practices: sc-best-practices.org and Science, 2024 [web:6][web:10][web:19]
"""

# Standard imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import nbinom, ttest_ind
from statsmodels.stats.multitest import multipletests
```

```python
# Jupyter Notebook Cell 2: Simulate Gene Expression Count Data

# Example apoptosis/healthy-relevant genes (from Nature): [web:1][web:5]
apoptosis_genes = [
    "CASP3", "CASP8", "BAX", "BCL2", "FAS", "FASLG", "XIAP", "TRADD", "TNFRSF10B",
    "JAK1", "JAK3", "STAT4", "IRF9", "CHMP1A", "CHMP5", "IFNAR1", "VDAC2"
]
# Add general core genes for more biological realism & dimension
core_genes = ["GAPDH", "ACTB", "RPL13A", "EEF1A1", "HPRT1"]
genes = core_genes + apoptosis_genes

n_genes = len(genes)
n_healthy = 20
n_apoptotic = 20

# Simulate counts: healthy baseline, apoptotic up-energy
np.random.seed(42)
counts_healthy = nbinom.rvs(20, 0.5, size=(n_genes, n_healthy))
# Upregulate apoptosis gene set in apoptotic samples:
apoptotic_fold_change = np.array([2 if g in apoptosis_genes else 1 for g in genes])
base_apoptotic = nbinom.rvs(20 * apoptotic_fold_change[:, None], 0.5, size=(n_genes, n_apoptotic))

# Combine and make DataFrame
counts = np.concatenate([counts_healthy, base_apoptotic], axis=1)
gene_df = pd.DataFrame(counts, index=genes, columns=[f"healthy_{i+1}" for i in range(n_healthy)] +
                                                     [f"apoptotic_{i+1}" for i in range(n_apoptotic)])
gene_df.head()
gene_df.to_csv('data.csv',index=True)
```

```python
import pandas as pd
gene_df = pd.read_csv('data.csv',index_col=0);
gene_df.head()
```

# Normalization and variance stabilization

- **To ensure that observed differences in gene expression reflect true biological variation, rather than technical bias or noise.**
- **Normalization ensures that differences in normalized read counts reflect genuine differences in gene expression**
- **Stabilization ensures highly expressed genes do not dominate analyses as genes with higher expression have higher variance.**


```python
# Normalization and Variance Stabilization

# Counts per Million normalization (Better approach TMM -- next time)
cpm = filtered_df.div(filtered_df.sum(axis=0), axis=1) * 1e6

# Variance stabilization (Better approach VST -- next time)
log_cpm = np.log1p(cpm)

# 
log_cpm.head()
```

```python
# Differential Expression Testing
import numpy as np
from scipy.stats import nbinom, ttest_ind
from statsmodels.stats.multitest import multipletests

# Generate label groupings
group_labels = np.array(["healthy"] * n_healthy + ["apoptotic"] * n_apoptotic)

# Check
pvals = []
logFC = []
for gene in filtered_df.index:
    grp_healthy = log_cpm.loc[gene, group_labels == "healthy"]
    grp_apoptotic = log_cpm.loc[gene, group_labels == "apoptotic"]
    t_stat, p = ttest_ind(grp_healthy, grp_apoptotic, equal_var=False)
    pvals.append(p)
    # Log2 fold change: apoptotic - healthy
    logFC.append(grp_apoptotic.mean() - grp_healthy.mean())
```

# P-value


```python
# Multiple testing correction (Benjamini-Hochberg FDR)
adj_pvals = multipletests(pvals, method="fdr_bh")[1]

# Results
results = pd.DataFrame({
    "gene": filtered_df.index,
    "log2FC": logFC,
    "pval": pvals,
    "padj": adj_pvals
}).set_index("gene")
results.head(20)
```

```python
#  Volcano Plot 

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
colors = []
for idx, row in results.iterrows():
    if row["log2FC"] > .2 and row["padj"] < 0.01:
        colors.append("red")
    elif row["log2FC"] < -.2 and row["padj"] < 0.01:
        colors.append("blue")
    else:
        colors.append("grey")
plt.scatter(results["log2FC"], -np.log10(results["padj"]), 
            c=colors, alpha=0.7)
plt.xlabel("Log2 fold change (apoptotic vs healthy)")
plt.ylabel("-log10 adjusted p-value")
plt.title("Volcano plot: Apoptosis DE Genes")
for gene in results.index:
    if abs(results.loc[gene, "log2FC"]) > 2 and results.loc[gene, "padj"] < 0.01:
        plt.text(results.loc[gene, "log2FC"], -np.log10(results.loc[gene, "padj"]), gene, fontsize=9)
plt.show()
```

```python
# Heatmap

# Sample color bar: healthy (green), apoptotic (magenta)
sample_type = pd.Series(["healthy"] * n_healthy + ["apoptotic"] * n_apoptotic, 
                        index=log_cpm.columns)
col_colors = sample_type.map({"healthy": "green", "apoptotic": "magenta"})

top_genes = results.sort_values("padj").head(10).index
data = log_cpm.loc[top_genes]

sns.clustermap(data, 
               cmap="vlag", 
               col_colors=col_colors,
               row_cluster=False, 
               col_cluster=False,
               xticklabels=False, 
               yticklabels=top_genes,
               cbar_kws={"label": "Log1p-CPM"},
               figsize=(10, 6))
plt.suptitle("Heatmap: Top 10 DE Genes (Green=Healthy, Magenta=Apoptotic)", y=1.05)
```
